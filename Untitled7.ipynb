```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("earthquake_1995-2023.csv")

# Count earthquakes by country
country_counts = df["country"].value_counts().head(10)

# Create bar plot
plt.figure(figsize=(10, 6))
country_counts.plot(kind="bar")

plt.title("Top 10 Countries by Number of Earthquakes")
plt.xlabel("Country")
plt.ylabel("Number of Earthquakes")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("earthquake_1995-2023.csv")

plt.figure(figsize=(10, 6))
plt.hist(df['magnitude'], bins=20, color='steelblue', edgecolor='black')
plt.xlabel('Magnitude')
plt.ylabel('Frequency')
plt.title('Histogram of Earthquake Magnitudes')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np

df = pd.read_csv("earthquake_1995-2023.csv")

# Bin magnitudes into ranges
bins = [6.5, 7, 7.5, 8, 8.5, 9, 9.5]
labels = ['6.5-7', '7-7.5', '7.5-8', '8-8.5', '8.5-9', '9-9.5']
df['mag_range'] = pd.cut(df['magnitude'], bins=bins, labels=labels, right=False)

mag_range_counts = df['mag_range'].value_counts().sort_index()

# Different color for each slice
colors = cm.plasma(np.linspace(0, 1, len(mag_range_counts)))

plt.figure(figsize=(8, 8))
plt.pie(
    mag_range_counts.values,
    labels=mag_range_counts.index.astype(str),
    autopct='%1.1f%%',   # show percentage on each slice
    startangle=90,
    colors=colors,
    explode=[0.03] * len(mag_range_counts)  # small gap between slices
)
plt.title('Distribution of Earthquakes by Magnitude Range')
plt.axis('equal')  # keeps the pie circular
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("earthquake_1995-2023.csv")

# Remove missing values
data = df[["magnitude", "depth"]].dropna()

# Create scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(data["depth"], data["magnitude"])

plt.title("Earthquake Magnitude vs Depth")
plt.xlabel("Depth (km)")
plt.ylabel("Magnitude")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("earthquake_1995-2023.csv")

# Convert date_time to datetime
df["date_time"] = pd.to_datetime(df["date_time"])

# Sort by date
df = df.sort_values("date_time")

# Create line plot
plt.figure(figsize=(12, 6))
plt.plot(df["date_time"], df["magnitude"])

plt.title("Earthquake Magnitude Over Time")
plt.xlabel("Date")
plt.ylabel("Magnitude")
plt.grid(True)
plt.tight_layout()
plt.show()
```
